```python

# Program Menghitung Total Harga Belanja di Kantin

nama_barang = input("Masukkan nama barang: ")
harga_barang = int(input("Masukkan harga barang: "))
jumlah = int(input("Masukkan jumlah yang dibeli: "))

total = harga_barang * jumlah

print()
print("Nama barang:", nama_barang)
print("Harga barang:", harga_barang)
print("Jumlah yang dibeli:", jumlah)
print("Total yang harus dibayar: Rp", total)
```

```python
nama = input("Masukkan nama siswa: ")
teori = float(input("Masukkan nilai teori: "))
praktek = float(input("Masukkan nilai praktek: "))

rata = (teori + praktek) / 2

if rata >= 75:
    status = "LULUS"
else:
    status = "TIDAK LULUS"

print("\nNama Siswa:", nama)
print("Rata-rata Nilai:", rata)
print("Status:", status)
```
